```python
import pandas as pd
```

```python
features_df = pd.read_csv("/content/features.csv", names =["features"])
labels_df = pd.read_csv("/content/labels.csv", names = ["labels"])

```

```python
data_df = pd.concat([features_df, labels_df], axis=1)

data_df.columns = ["features", "labels"]

data_df = data_df.reset_index().rename(columns={"index": "id"})
```

```python
freq = data_df["labels"].value_counts()

print(freq)
```

```python
ratios = {
    "gambiae": 1473,
    "coluzzii": 705,
    "arabiensis": 368,
    "bissau": 170,
    "gcx3": 65
}

total = sum(ratios.values())

samples = {k: int(v / total * 100) for k, v in ratios.items()}

# adjust to ensure total = 100
diff = 100 - sum(samples.values())
for k in list(samples.keys())[:diff]:
    samples[k] += 1

sampled = []

for label, n in samples.items():
    sampled.append(data_df[data_df["labels"] == label].sample(n=n, random_state=42))

result = pd.concat(sampled).sample(frac=1, random_state=42)  # shuffle final

result
```

```python
freq_new = result["labels"].value_counts()

print(freq_new)
```

```python
data_df.to_csv("data.csv", index=False)
```
